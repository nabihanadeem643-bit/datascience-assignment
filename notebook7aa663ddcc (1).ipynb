```python
import pandas as pd

df = pd.read_csv('/kaggle/input/competitions/titanic/train.csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Make a copy of the original dataset
clean_df = df.copy()

# Create a feature showing whether cabin information exists
clean_df['HasCabin'] = clean_df['Cabin'].notna().astype(int)

# Fill missing Age values with the median
clean_df['Age'] = clean_df['Age'].fillna(clean_df['Age'].median())

# Fill missing Embarked values with the mode
clean_df['Embarked'] = clean_df['Embarked'].fillna(
    clean_df['Embarked'].mode()[0]
)

# Drop the original Cabin column
clean_df = clean_df.drop(columns=['Cabin'])

# Check missing values after cleaning
print("Missing values after cleaning:")
print(clean_df.isnull().sum())

print("\nNew dataset shape:", clean_df.shape)
```
